```python
# split data == 
# training == 80% x_train, y_train , x y
# Testing  == 20% x_test, y_test
```

```python
# train_test_split
# sklearn     , (siketlearn)
from sklearn.model_selection import train_test_split      
# sklearn = package , model_selection =file , train_test_split function
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)

# x_train =80% , x_test = 20%, y_train = 80% ,y_test = 20%        x = input, y = output
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
# metrics IT TELLS PERFORMANCE OF MODEL
# accuracy, mse(mean sqaure error), mae(mean absoulte error) trade of Variance, bias
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
y_pred = model.predict(x_test)
```

```python
y_test
```

```python
y_pred.shape
```

```python
# testing accuracy
```

```python
accuracy = r2_score(y_test, y_pred) 
```

```python
accuracy*100
```

```python
# Training
r2_score(y_train, model.predict(x_train))*100
```

```python
# metrics
```

```python
plt.scatter(x_train, y_train, color = "red", label = "Training Data")
plt.scatter(x_test, y_test, color = "green", label = "Testing Data")
plt.plot(x_test, y_pred, color = "blue", label = "BFT")
plt.legend()
plt.title("BFT for Training Data")
plt.show()
```

```python
df.head(3)  
```

```python
# Bias / Error
# Actual Output -  Predict Output

pd.DataFrame(
    {"Actual Output" : y_test,
    "Predicted Output" : y_pred,
    "Bias/Error" : y_test - y_pred,
    "Error in % ": (((y_test - y_pred)/y_test)*100 ).round(2)
    }
)
```

```python
plt.scatter(x_train, y_train, color = "red", label = "Training Data")
plt.scatter(x_test, y_test, color = "green", label = "Testing Data")
plt.plot(x_train, model.predict(x_train), color = "blue", label = "BFT")
plt.legend()
plt.title("BFT for Training Data")
plt.show()
```
